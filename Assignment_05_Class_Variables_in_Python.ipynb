#Assignment-5: Class Variables in Python

1. Define and Access via Class

```python
class Student:
    college = "CMREC"   # Class variable

print("College:", Student.college)
```

2. Access via Object (Instance)

```python
class Student:
    college = "CMREC"

s1 = Student()

print("College:", s1.college)
```

3. Modify Class Variable via Instance

```python
class Student:
    college = "CMR"

s1 = Student()

# Creates an instance variable, does not change class variable
s1.college = "CMR Engineering College"

print("Using Instance:", s1.college)
print("Using Class:", Student.college)
```

4. Modify Class Variable via Class

```python
class Student:
    college = "JNTUH"
s1 = Student()
s2 = Student()
Student.college = "CMR Engineering College"
print("Using Class:", Student.college)
print("Using s1:", s1.college)
print("Using s2:", s2.college)
```

5. Class Variable vs Instance Variable

```python
class Student:
    college = "JNTUH"   # Class variable

    def __init__(self, name):
        self.name = name       # Instance variable

s1 = Student("Sathvika")
s2 = Student("Ash")

print("Before Modification")
print("s1 Name:", s1.name)
print("s2 Name:", s2.name)
print("College:", Student.college)

# Modify instance variable
s1.name = "Anjali"

# Modify class variable
Student.college = "CMR Engineering College"

print("\nAfter Modification")
print("s1 Name:", s1.name)
print("s2 Name:", s2.name)
print("s1 College:", s1.college)
print("s2 College:", s2.college)
```

6. Shared Counter Program

```python
class Student:
    count = 0
    def __init__(self, name):
        self.name = name
        Student.count += 1
s1 = Student("Sathvika")
s2 = Student("Rahul")
s3 = Student("Anjali")
print("Total Students:", Student.count)
```

7. Configuration Setting Example

```python
class Course:
    course_name = "Python Basics"
c1 = Course()
c2 = Course()
print("Before Modification")
print("c1:", c1.course_name)
print("c2:", c2.course_name)
Course.course_name = "Advanced Python"
print("\nAfter Modification")
print("c1:", c1.course_name)
print("c2:", c2.course_name)
print("Class:", Course.course_name)
```
