#Assignment-10: Packages & Modules in Python

1. Create Two Classes in Separate Files

class_one.py


```python
%%writefile class_one.py

class ClassOne:
    def __init__(self, name):
        self.name = name
        print("ClassOne Constructor Executed")

    def display(self):
        print("Name:", self.name)
```

class_two.py

```python
%%writefile mypackage/class_two.py

from mypackage.class_one import ClassOne

class ClassTwo:
    def __init__(self, course):
        self.course = course
        print("ClassTwo Constructor Executed")

    def display_course(self):
        print("Course:", self.course)
    def create_object(self):
        obj = ClassOne("Relative Import User")
        obj.display()
```

2: Create a Package Folder

```python
!mkdir -p mypackage
!mv class_one.py mypackage/
!mv class_two.py mypackage/
```

```python
%%writefile mypackage/__init__.py

from .class_one import ClassOne
from .class_two import ClassTwo
```

3: Import Classes

```python
import mypackage.class_one
from mypackage.class_two import ClassTwo
```

4. Use the Classes

```python
obj1 = mypackage.class_one.ClassOne("Sathvika")
obj1.display()

obj2 = ClassTwo("Python")
obj2.display_course()
```

5. Final Program

```python
%%writefile main.py
import mypackage.class_one
from mypackage.class_two import ClassTwo
print("Using ClassOne")
obj1 = mypackage.class_one.ClassOne("Sathvika")
obj1.display()
print()
print("Using ClassTwo")
obj2 = ClassTwo("Python")
obj2.display_course()
```

```python
!python main.py
```

6. Alias Imports

```python
import mypackage.class_one as c1

obj = c1.ClassOne("Student")
obj.display()
```

7. Relative Imports

```python
%%writefile mypackage/class_two.py

from mypackage.class_one import ClassOne

class ClassTwo:
    def __init__(self, course):
        self.course = course
        print("ClassTwo Constructor Executed")

    def display_course(self):
        print("Course:", self.course)

    def create_object(self):
        obj = ClassOne("Relative Import User")
        obj.display()
```

```python
import importlib
import mypackage.class_two

importlib.reload(mypackage.class_two)

from mypackage.class_two import ClassTwo

obj = ClassTwo("Python")
obj.create_object()
```
