#Assignment-15: Dictionaries in Python

1. Create a Dictionary

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}

print("Student Dictionary:")
print(students)
```

1.1 Add New Entries

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan"
}
students[104] = "Divya"
students[105] = "Esha"
print(students)
```

1.2 Update Existing Values

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan"
}
students[102] = "Bhavana"
print(students)
```

1.3 Access Values

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}
print("Student with ID 103:", students[103])
print("All Student Names:")
print(list(students.values()))
```

1.4 Iterate Through Dictionary

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}

for key, value in students.items():
    print(key, ":", value)
```

1.5 Print Only Keys

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}
print("Student IDs:")
for key in students.keys():
    print(key)
```

1.6 Print Only Values

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}

print("Student Names:")
for value in students.values():
    print(value)
```

1.7 Create a Nested Dictionary

```python
students = {
    101: {"name": "Aarav", "age": 20},
    102: {"name": "Bhavya", "age": 21},
    103: {"name": "Charan", "age": 22}
}

print(students)
```

1.8 Access Nested Values

```python
students = {
    101: {"name": "Aarav", "age": 20},
    102: {"name": "Bhavya", "age": 21},
    103: {"name": "Charan", "age": 22}
}
print("Student Name:", students[102]["name"])
print("Student Age :", students[102]["age"])
```

1.9 Delete Elements

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}
del students[101]
print("After del:", students)
students.popitem()
print("After popitem():", students)
students.pop(103)
print("After pop():", students)
```

2. Check Key Existence

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan"
}
student_id = 102
if student_id in students:
    print("Student ID exists.")
else:
    print("Student ID does not exist.")
```

3. Count Entries

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan",
    104: "Divya",
    105: "Esha"
}

print("Total Students:", len(students))
```

4. Merge Two Dictionaries

```python
students1 = {
    101: "Aarav",
    102: "Bhavya"
}
students2 = {
    103: "Charan",
    104: "Divya"
}
merged = students1 | students2
print("Merged Dictionary:")
print(merged)
```

5. Dictionary Comprehension

```python
squares = {num: num**2 for num in range(1, 6)}
print(squares)
```

6. Reverse Dictionary

```python
students = {
    101: "Aarav",
    102: "Bhavya",
    103: "Charan"
}
reverse = {value: key for key, value in students.items()}
print(reverse)
```
