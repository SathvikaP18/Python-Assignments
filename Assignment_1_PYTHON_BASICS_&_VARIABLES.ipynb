3. Working with Basic Data Types

```python
# Integer
num = 10

# Float
price = 99.99

# Boolean
is_student = True

# String (single character)
grade = "A"

print("Integer:", num, "Type:", type(num))
print("Float:", price, "Type:", type(price))
print("Boolean:", is_student, "Type:", type(is_student))
print("String:", grade, "Type:", type(grade))
```

4. Local vs Global Variables

```python
# Global variable
name = "Global Variable"

def demo():
    global name

    # Local variable
    local_name = "Local Variable"

    print("Local Variable:", local_name)
    print("Global Variable:", name)

    # Modify global variable
    name = "Modified Global Variable"

demo()

print("Global Variable after function call:", name)
```

5. Type Checking & Dynamic Typing

```python
x = 10
print("Value:", x)
print("Type:", type(x))

x = 25.5
print("Value:", x)
print("Type:", type(x))

x = True
print("Value:", x)
print("Type:", type(x))

x = "Python"
print("Value:", x)
print("Type:", type(x))
```
