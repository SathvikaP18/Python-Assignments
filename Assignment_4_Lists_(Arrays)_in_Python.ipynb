#Assignment-4: Lists (Arrays) in Python

1. Sum of Elements

```python
def sum_list(lst):
    total = 0
    for i in lst:
        total += i
    return total
numbers = [10, 20, 30, 40, 50]
print("Sum =", sum_list(numbers))
```

2. Average of Elements

```python
def average(lst):
    total = 0
    for i in lst:
        total += i
    return total / len(lst)
numbers = [10, 20, 30, 40, 50]
print("Average =", average(numbers))
```

3. Find Index of an Element

```python
numbers = [10, 20, 30, 40, 50]
element = int(input("Enter element: "))
if element in numbers:
    print("Index =", numbers.index(element))
else:
    print("Element not found.")
```

4. Check Element Presence

```python
def check_element(lst, value):
    if value in lst:
        print("Element Found")
    else:
        print("Element Not Found")
numbers = [10, 20, 30, 40]
check_element(numbers, 20)
```

5. Remove an Element

```python
def remove_element(lst, value):
    if value in lst:
        lst.remove(value)
    return lst
numbers = [10, 20, 30, 40]
print(remove_element(numbers, 30))
```

6. Copy a List

```python
def copy_list(lst):
    new_list = []
    for i in lst:
        new_list.append(i)
    return new_list
numbers = [10, 20, 30]
print(copy_list(numbers))
```

7. Insert Element at Position

```python
def insert_element(lst, index, value):
    lst.insert(index, value)
    return lst
numbers = [10, 20, 40]
print(insert_element(numbers, 2, 30))
```

8. Find Minimum and Maximum

```python
def min_max(lst):
    print("Minimum =", min(lst))
    print("Maximum =", max(lst))
numbers = list(map(int, input("Enter numbers: ").split(",")))
min_max(numbers)
```

9. Reverse a List

```python
def reverse_list(lst):
    lst.reverse()
    return lst
numbers = [10, 20, 30, 40]
print(reverse_list(numbers))
```

10. Find Duplicate Elements

```python
def find_duplicates(lst):
    duplicates = []
    for i in lst:
        if lst.count(i) > 1 and i not in duplicates:
            duplicates.append(i)
    return duplicates
numbers = [1, 2, 3, 2, 4, 5, 1]
print(find_duplicates(numbers))
```

11. Count Even and Odd Numbers

```python
def count_even_odd(lst):
    even = 0
    odd = 0
    for i in lst:
        if i % 2 == 0:
            even += 1
        else:
            odd += 1
    print("Even =", even)
    print("Odd =", odd)
numbers = [1, 2, 3, 4, 5, 6]
count_even_odd(numbers)
```

12. Common Elements Between Two Lists

```python
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
common = []
for i in list1:
    if i in list2:
        common.append(i)
print(common)
```

13. Remove Duplicates

```python
def remove_duplicates(lst):
    unique = []
    for i in lst:
        if i not in unique:
            unique.append(i)
    return unique
numbers = [1, 2, 2, 3, 4, 4, 5]
print(remove_duplicates(numbers))
```

14. Second Largest Element

```python
def second_largest(lst):
    unique = []
    for i in lst:
        if i not in unique:
            unique.append(i)
    unique.sort()
    print("Second Largest =", unique[-2])
numbers = [10, 50, 20, 40, 50]
second_largest(numbers)
```

15. Difference Between Max and Min

```python
def difference(lst):
    return max(lst) - min(lst)
numbers = [10, 30, 5, 50]
print("Difference =", difference(numbers))
```

16. Check for Specific Elements

```python
def check_values(lst):
    if 12 in lst and 23 in lst:
        print("Both 12 and 23 are present.")
    else:
        print("Both values are not present.")
numbers = [5, 12, 18, 23]
check_values(numbers)
```

17. Unique Elements Only

```python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique = []
for i in numbers:
    if i not in unique:
        unique.append(i)
print(unique)
```

18. Frequency Count

```python
numbers = [1, 2, 2, 3, 3, 3, 4]
for i in set(numbers):
    print(i, ":", numbers.count(i))
```

19. List Sorting Without Built-in Functions

```python
numbers = [30, 10, 50, 20, 40]
for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] > numbers[j]:
            numbers[i], numbers[j] = numbers[j], numbers[i]
print(numbers)
```

20. Merge Two Lists Without Duplicates

```python
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
merged = list1 + list2
result = []
for i in merged:
    if i not in result:
        result.append(i)
print(result)
```
