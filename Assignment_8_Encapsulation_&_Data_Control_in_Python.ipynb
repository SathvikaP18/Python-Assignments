1. Create a Simple Class

```python
class Student:
    def set_values(self, name, age):
        self.name = name
        self.age = age

    def get_values(self):
        print("Name:", self.name)
        print("Age:", self.age)


student = Student()
student.set_values("Sathvika", 20)
student.get_values()
```

2. Validation Using Methods

```python
class Student:
    def set_values(self, name, age):
        self.name = name

        if age > 0:
            self.age = age
        else:
            print("Error: Age should be greater than 0.")
            self.age = None

    def get_values(self):
        print("Name:", self.name)
        print("Age:", self.age)


student = Student()
student.set_values("Sathvika", 20)
student.get_values()

student.set_values("Ashwini", -5)
student.get_values()
```

3. Property Decorator

```python
class Student:
    def __init__(self):
        self._name = ""
        self._age = 0
    @property
    def name(self):
        return self._name
    @name.setter
    def name(self, value):
        self._name = value

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value > 0:
            self._age = value
        else:
            print("Error: Age should be greater than 0.")

student = Student()

student.name = "Sathvika"
student.age = 20

print("Name:", student.name)
print("Age:", student.age)

student.age = -2
```

4. Read-Only Property

```python
class Employee:
    def __init__(self, emp_id):
        self._id = emp_id

    @property
    def id(self):
        return self._id


emp = Employee(101)

print("Employee ID:", emp.id)
```

5. Bank Account System

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance.")

    def show_balance(self):
        print("Current Balance:", self.balance)


account = BankAccount("Sathvika", 5000)

account.deposit(1000)
account.show_balance()

account.withdraw(2000)
account.show_balance()

account.withdraw(5000)
```

6. Internal Variables

7. Computed Property

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    @property
    def area(self):
        return self.length * self.width

rect = Rectangle(10, 5)

print("Length:", rect.length)
print("Width:", rect.width)
print("Area:", rect.area)
```

8. Password Validation System

```python
class User:
    def __init__(self):
        self._password = ""

    @property
    def password(self):
        return "*" * len(self._password)

    @password.setter
    def password(self, value):
        if len(value) < 8:
            print("Password must be at least 8 characters.")
        elif not any(ch.isdigit() for ch in value):
            print("Password must contain at least one number.")
        else:
            self._password = value
            print("Password set successfully.")


user = User()

user.password = "python"
user.password = "python12"

print("Stored Password:", user.password)
```
