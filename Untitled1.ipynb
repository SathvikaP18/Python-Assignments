2. Define Methods

```python
class A:
    def method1(self):
        print("A - Method 1")
    def method2(self):
        print("A - Method 2")
    def common_method(self):
        print("Common Method from Class A")

class B(A):
    def method3(self):
        print("B - Method 1")
    def method4(self):
        print("B - Method 2")
    def common_method(self):
        print("Common Method from Class B")

class C(B):
    def method5(self):
        print("C - Method 1")

    def method6(self):
        print("C - Method 2")

    def common_method(self):
        print("Common Method from Class C")
```

3. Object Creation and Method Calls

```python
class A:
    def method1(self):
        print("A - Method 1")

    def method2(self):
        print("A - Method 2")

    def common_method(self):
        print("Common Method from A")


class B(A):
    def method3(self):
        print("B - Method 1")

    def method4(self):
        print("B - Method 2")

    def common_method(self):
        print("Common Method from B")


class C(B):
    def method5(self):
        print("C - Method 1")

    def method6(self):
        print("C - Method 2")

    def common_method(self):
        print("Common Method from C")


a = A()
b = B()
c = C()

print("Class A")
a.method1()
a.method2()
a.common_method()

print("\nClass B")
b.method1()
b.method2()
b.method3()
b.method4()
b.common_method()

print("\nClass C")
c.method1()
c.method2()
c.method3()
c.method4()
c.method5()
c.method6()
c.common_method()
```

4. Superclass Reference Behavior (Runtime Polymorphism)

```python
class A:
    def common_method(self):
        print("Common Method from A")

class B(A):
    def common_method(self):
        print("Common Method from B")

class C(B):
    def common_method(self):
        print("Common Method from C")

ref = B()
ref.common_method()

ref = C()
ref.common_method()
```

5. Access Parent Method using super()

```python
class A:
    def common_method(self):
        print("Common Method from A")

class B(A):
    def common_method(self):
        super().common_method()
        print("Common Method from B")

class C(B):
    def common_method(self):
        super().common_method()
        print("Common Method from C")

obj = C()
obj.common_method()
```

6. Instance Variables Behavior

```python
class A:
    def __init__(self):
        self.name = "Class A"

class B(A):
    def __init__(self):
        super().__init__()
        self.name = "Class B"

class C(B):
    def __init__(self):
        super().__init__()
        self.name = "Class C"

a = A()
b = B()
c = C()

print("Object A:", a.name)
print("Object B:", b.name)
print("Object C:", c.name)

ref = B()
print("Superclass Reference to B:", ref.name)

ref = C()
print("Superclass Reference to C:", ref.name)
```

8. Real-World Example (Vehicle → Car → ElectricCar)

```python
class Vehicle:
    def start_engine(self):
        print("Vehicle engine started")

    def stop_engine(self):
        print("Vehicle engine stopped")

class Car(Vehicle):
    def start_engine(self):
        super().start_engine()
        print("Car engine started with key")

    def drive(self):
        print("Car is driving")

class ElectricCar(Car):
    def start_engine(self):
        super().start_engine()
        print("Electric Car starts silently")

    def charge(self):
        print("Electric Car is charging")

ev = ElectricCar()

ev.start_engine()
ev.drive()
ev.charge()
ev.stop_engine()
```
